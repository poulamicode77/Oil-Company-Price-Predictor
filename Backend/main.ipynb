```python
from huggingface_hub import InferenceClient

# 1. Setup the connection
# Replace YOUR_API_KEY_HERE with your actual token
client = InferenceClient(api_key= "ggdjsj")

# 2. Ask the question
messages = [
    {
        "role": "user",
        "content": "Hello! I am building an oil price bot. Tell me one fun fact about oil."
    }
]

# 3. Get the response
# Using Qwen2.5 because it is officially supported as a 'Chat' model
try:
    response = client.chat.completions.create(
        model="Qwen/Qwen2.5-7B-Instruct",
        messages=messages,
        max_tokens=500,
    )
    print(response.choices[0].message.content)
except Exception as e:
    print(f"Still having an issue? Here is the error: {e}")
```

```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# 1. Our "Company Database" (The Internal Data)
company_knowledge = [
    "Oil Project Alpha: Located in the North Sea. Production cost: $45 per barrel.",
    "Logistics Report: Shipping to India takes 15 days and costs $5 per barrel.",
    "Strategic Plan: We aim to keep prices 10% above production costs.",
    "Inventory: Current storage in Rotterdam is at 85% capacity.",
    "Personnel: Chief Analyst is Dr. Sarah Miller, specialized in OPEC trends."
]

# 2. Load a "Map Maker" (Embedding Model)
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Turn our sentences into numbers (Embeddings)
embeddings = model.encode(company_knowledge)

# 4. Create the "Library" (Vector Database)
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))

print("Step 2 Complete: Your company library is ready and digitized!")
```

```python
# The User's Question
question = "How long does it take to ship oil to India?"

# Turn the question into numbers
questiion_embedding = model.encode([question])

# Search the library for the closest match (k=1 means give me the top 1 match)
distances, indices = index.search(np.array(questiion_embedding), k=1)

# Print the result
print(f"Question: {question}")
print(f"Retrieved Info: {company_knowledge[indices[0][0]]}")
```
